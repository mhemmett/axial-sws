```python
# Import requirements
import pandas as pd
from get_all_traces import *
import matplotlib.pyplot as plt
```

```python
# Load in catalog with date_time for pulling IRIS waveforms
sample_catalog = pd.read_csv('../data/old_catalog_with_geometry_filtered.csv')
```

```python
sample_catalog = sample_catalog.head(100)
```

```python
#sample_catalog.to_csv('../data/old_catalog_with_geometry_filtered_mini.csv', index=False)
```

```python
display(sample_catalog)
```

```python
# Convert datetime column to datetime type if it's a string
sample_catalog['datetime'] = pd.to_datetime(sample_catalog['datetime'])
# Add 15 seconds to datetime to create endtime
sample_catalog['endtime'] = sample_catalog['datetime'] + pd.Timedelta(seconds=15)
```

```python
sample_catalog = sample_catalog.drop(['month', 'day', 'hour','minute', 'second', 'event_datetime', 'weight', 
                                      'datetime_str', 'event_lat', 'event_lon', 'event_depth', 'quality',
                                      'line_number_phase'], axis=1)
```

```python
sample_catalog = sample_catalog.drop(['line_number_event', 'raw_line_event'], axis=1)
```

```python
sample_catalog = sample_catalog.drop(['raw_line_phase'], axis=1)
```

```python
display(sample_catalog)
```

```python
catalog = sample_catalog
```

```python
# For each event, with phase_type 'P', save arrival_time as 'p_arrival_time' and with phase_type 'S', save arrival_time as 's_arrival_time'
p_arrivals = catalog[catalog['phase_type'] == 'P'][['event_id', 'arrival_time']].rename(columns={'arrival_time': 'p_arrival_time'})
s_arrivals = catalog[catalog['phase_type'] == 'S'][['event_id', 'arrival_time']].rename(columns={'arrival_time': 's_arrival_time'})

# Merge P and S arrivals back into the main catalog
catalog = catalog.merge(p_arrivals, on='event_id', how='left').merge(s_arrivals, on='event_id', how='left')

# Remove duplicates based on id and station_id, keeping the first occurrence
catalog_final = catalog.drop_duplicates(subset=['event_id', 'station_id'])
```

```python
catalog_final = catalog_final.drop(['phase_type', 'arrival_time'], axis=1)
```

```python
display(catalog_final)
```

```python
get_all_traces(catalog_final, 'old_catalog_waveforms_100', 'datetime', 'endtime')
```
