```python
# Let's edit these functions to plot the rose plots with only the higheest bin and the two adjacent bins, to better visualize the dominant splitting direction and its uncertainty.
# We will print them in black, with a circular outline to indicate N, E, S, W directions.
# Otherwise, no background so it can be easily plotted on top of a map of the volcano

def plot_fast_direction_rose_eruption_comparison(results_df, eruption_time=None, 
                                                  title_prefix="Fast Direction Distribution",
                                                  nbins=36, figsize=(16, 7), color='black',
                                                  edgecolor='black', linewidth=0.5,
                                                  time_column='event_datetime'):
    """
    Create side-by-side 360° rose plots comparing fast directions before and after eruption.
    
    Parameters:
    -----------
    results_df : pd.DataFrame
        DataFrame with splitting results containing 'phi' and time columns
    eruption_time : UTCDateTime or str
        Time of eruption onset (default: 2015-04-24T06:00:00)
    title_prefix : str
        Prefix for plot titles
    nbins : int
        Number of angular bins (default 36 = 10° bins for 360°)
    figsize : tuple
        Figure size (width, height) for combined plot
    color : str
        Color for the histogram bars
    edgecolor : str
        Color for bar edges
    linewidth : float
        Width of bar edges
    time_column : str
        Name of the time column in results_df (default 'event_datetime')
    """
    if eruption_time is None:
        eruption_time = UTCDateTime(2015, 4, 24, 6)
    elif not isinstance(eruption_time, UTCDateTime):
        eruption_time = UTCDateTime(eruption_time)
    
    # Convert time column to UTCDateTime for comparison
    df_time = results_df[time_column].apply(lambda x: UTCDateTime(x) if not isinstance(x, UTCDateTime) else x)
    results_before = results_df[df_time < eruption_time]
    results_after = results_df[df_time >= eruption_time]
    
    # Create figure with two subplots
    fig = plt.figure(figsize=figsize)
    
    # Before eruption plot (left)
    ax1 = fig.add_subplot(121, projection='polar')
    plot_rose_subplot(results_before, ax1, f"{title_prefix}\nBefore Eruption", 
                      nbins, color, edgecolor, linewidth)
    
    # After eruption plot (right)
    ax2 = fig.add_subplot(122, projection='polar')
    plot_rose_subplot(results_after, ax2, f"{title_prefix}\nAfter Eruption", 
                      nbins, color, edgecolor, linewidth)
    
    plt.tight_layout()
    return fig, (ax1, ax2), (results_before, results_after)


def plot_rose_subplot(results_df, ax, title, nbins, color, edgecolor, linewidth):
    """
    Helper function to plot rose diagram on a given axis.
    """
    if len(results_df) == 0:
        ax.text(0.5, 0.5, 'No data', transform=ax.transAxes, 
                ha='center', va='center', fontsize=14)
        return
    
    fast_directions = []
    for phi in results_df['phi'].values:  # in degrees (-90 to +90)
        # Convert to 0-360 range and add 180° symmetry
        phi_0_180 = phi + 90  # Convert to 0-180 range
        phi_rad_1 = np.deg2rad(phi_0_180)
        phi_rad_2 = np.deg2rad(phi_0_180 + 180)  # Add 180° symmetric value
        fast_directions.append(phi_rad_1)
        fast_directions.append(phi_rad_2)
    
    fast_directions = np.array(fast_directions)
    
    # Create histogram bins (0 to 2π for 0° to 360°)
    bins = np.linspace(0, 2*np.pi, nbins + 1)
    
    # Calculate histogram
    counts, bin_edges = np.histogram(fast_directions, bins=bins)
    
    # Calculate bin centers
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
    
    # Width of each bar
    width = 2 * np.pi / nbins
    
    # Create the rose plot
    bars = ax.bar(bin_centers, counts, width=width, bottom=0.0,
                   color=color, edgecolor=edgecolor, linewidth=linewidth, alpha=0.7)
    
    # Set theta direction (clockwise from North)
    ax.set_theta_zero_location('N')
    ax.set_theta_direction(-1)
    
    # Set radial ticks
    ax.set_rlabel_position(45)
    
    # Add degree labels for full 360°
    tick_labels = ['0°', '45°', '90°', '135°', '180°', '225°', '270°', '315°']
    tick_positions = np.deg2rad([0, 45, 90, 135, 180, 225, 270, 315])
    ax.set_xticks(tick_positions)
    ax.set_xticklabels(tick_labels)
    
    # Adjust y-limit
    max_count = counts.max()
    ax.set_ylim(0, max_count * 1.2)
    
    # Add title with statistics
    n_measurements = len(fast_directions) // 2  # Divide by 2 since we doubled for symmetry
    # Calculate circular mean
    original_directions = np.deg2rad(results_df['phi'].values)
    mean_direction = np.rad2deg(np.arctan2(np.sin(original_directions).sum(), 
                                           np.cos(original_directions).sum()))
    
    title_text = f"{title}\nN = {n_measurements}, Mean = {mean_direction:.1f}°"
    ax.set_title(title_text, va='bottom', fontsize=12, fontweight='bold', pad=20)
    
    # Add grid
    ax.grid(True, linestyle='--', alpha=0.5)

    

```

```python
def plot_fast_direction_rose_eruption_comparison(results_df, eruption_time=None, output_filename=None,
                                                  title_prefix="Fast Direction Distribution",
                                                  nbins=36, figsize=(16, 7), color='black',
                                                  edgecolor='black', linewidth=0.5,
                                                  time_column='event_datetime'):
    if eruption_time is None:
        eruption_time = UTCDateTime(2015, 4, 24, 6)
    elif not isinstance(eruption_time, UTCDateTime):
        eruption_time = UTCDateTime(eruption_time)

    df_time = results_df[time_column].apply(
        lambda x: UTCDateTime(x) if not isinstance(x, UTCDateTime) else x
    )
    results_before = results_df[df_time < eruption_time]
    results_after  = results_df[df_time >= eruption_time]

    fig = plt.figure(figsize=figsize)

    ax1 = fig.add_subplot(121, projection='polar')
    plot_rose_subplot(results_before, ax1, f"{title_prefix}\nBefore Eruption",
                      nbins, color, edgecolor, linewidth)

    ax2 = fig.add_subplot(122, projection='polar')
    plot_rose_subplot(results_after, ax2, f"{title_prefix}\nAfter Eruption",
                      nbins, color, edgecolor, linewidth)

    plt.tight_layout()
    fig.patch.set_alpha(0.0)

    if output_filename:
        fig.savefig(output_filename, transparent=True, bbox_inches='tight')
    return fig, (ax1, ax2), (results_before, results_after)


def plot_rose_subplot(results_df, ax, title, nbins, color, edgecolor, linewidth):
    """
    Rose diagram showing only the dominant bin and its two immediate neighbours,
    with a transparent background and a plain circular NESW compass outline.
    """
    if len(results_df) == 0:
        ax.text(0.5, 0.5, 'No data', transform=ax.transAxes,
                ha='center', va='center', fontsize=14)
        return

    # ------------------------------------------------------------------
    # Build the full 360° histogram (doubling for axial symmetry as before)
    # ------------------------------------------------------------------
    fast_directions = []
    for phi in results_df['phi'].values:
        phi_0_180 = phi + 90
        fast_directions.append(np.deg2rad(phi_0_180))
        fast_directions.append(np.deg2rad(phi_0_180 + 180))
    fast_directions = np.array(fast_directions)

    bins        = np.linspace(0, 2 * np.pi, nbins + 1)
    counts, bin_edges = np.histogram(fast_directions, bins=bins)
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
    width       = 2 * np.pi / nbins

    # ------------------------------------------------------------------
    # Identify the peak bin and its two immediate neighbours (wrap-around)
    # ------------------------------------------------------------------
    peak_idx     = np.argmax(counts)
    antipodal_idx = (peak_idx + nbins // 2) % nbins
    keep_indices = {
        (peak_idx - 1) % nbins,
        peak_idx,
        (peak_idx + 1) % nbins,
        (antipodal_idx - 1) % nbins,
        antipodal_idx,
        (antipodal_idx + 1) % nbins,
    }

    # Build masked count array — zero everywhere except the three kept bins
    display_counts = np.zeros_like(counts, dtype=float)
    for i in keep_indices:
        display_counts[i] = counts[i]

    # ------------------------------------------------------------------
    # Draw only the three bars
    # ------------------------------------------------------------------
    ax.bar(bin_centers, display_counts, width=width, bottom=0.0,
           color=color, edgecolor=edgecolor, linewidth=linewidth,
           alpha=1.0)

    # ------------------------------------------------------------------
    # Transparent background
    # ------------------------------------------------------------------
    ax.set_facecolor('none')
    fig = ax.get_figure()
    fig.patch.set_alpha(0.0)

    # ------------------------------------------------------------------
    # Compass outline: simple circle + N/E/S/W labels, no grid/ticks/spines
    # ------------------------------------------------------------------
    ax.set_theta_zero_location('N')
    ax.set_theta_direction(-1)

    # Hide all default polar decorations
    ax.set_yticklabels([])
    ax.set_xticklabels([])
    ax.set_xticks([])
    ax.yaxis.set_visible(False)
    ax.spines['polar'].set_visible(True)
    ax.spines['polar'].set_linewidth(1.5)
    ax.spines['polar'].set_color('black')
    ax.grid(False)

    # Radial limit — scale to the full histogram so bar heights are meaningful
    max_count = counts.max()
    ax.set_ylim(0, max_count * 1.2)

    # Cardinal labels placed just outside the circle
    r_label = max_count * 1.28
    for label, angle_deg in [('N', 0), ('E', 90), ('S', 180), ('W', 270)]:
        ax.text(np.deg2rad(angle_deg), r_label, label,
                ha='center', va='center',
                fontsize=9, fontweight='bold', color='black')

    # ------------------------------------------------------------------
    # Title with N and circular mean
    # ------------------------------------------------------------------
    n_measurements  = len(fast_directions) // 2
    original_rad    = np.deg2rad(results_df['phi'].values)
    mean_direction  = np.rad2deg(
        np.arctan2(np.sin(original_rad).sum(), np.cos(original_rad).sum())
    )
    #ax.set_title(f"{title}\nN={n_measurements},  mean={mean_direction:.1f}°",
    #             va='bottom', fontsize=11, fontweight='bold', pad=20, color='black')
```
