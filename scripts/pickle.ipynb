```python
# Load ./SWS_nlloc/split_001.nlloc with obspy and save just the first event to a new nlloc file
import obspy

nlloc_file = './SWS_nlloc/split_001.nlloc'
cat = obspy.read_events(nlloc_file)
new_cat = obspy.core.event.Catalog(events=[cat.events[0]])
new_nlloc_file = './SWS_nlloc/split_001_event1.nlloc'
new_cat.write(new_nlloc_file, format='NLLOC_OBS')
```
