# Identify where Seismic Stations OO.AXID1 and OO.AXCC1 tilted during the 2015 eruption of Axial Seamount
Print out seismograms before, during, and after the eruption to identify when the tilting started and when the instruments flattened out again

```python
import warnings
from obspy import UTCDateTime
from obspy.clients.fdsn import Client
import matplotlib.pyplot as plt

client = Client("IRIS")
```

```python

network = "OO"
stations = ["AXCC1"]

start = UTCDateTime("2015-04-28T00:00:00")
end = UTCDateTime("2015-04-29T00:00:00")
window_seconds = 5 * 60  # 5 minutes

# Prefer broadband-like bands in this order
band_priority = ["HH", "EH"]

def pick_band_and_loc(inventory):
    """
    Return (location, band) for a complete Z/N/E triplet.
    """
    available = {}
    for net in inventory:
        for sta in net:
            for ch in sta.channels:
                if len(ch.code) == 3 and ch.code[-1] in "ZNE":
                    available.setdefault(ch.location_code, set()).add(ch.code)

    for band in band_priority:
        needed = {f"{band}Z", f"{band}N", f"{band}E"}
        for loc, chans in sorted(available.items()):
            if needed.issubset(chans):
                return loc, band
    return None, None

t0 = start
while t0 < end:
    t1 = t0 + window_seconds
    print(f"\n===== Week window starting {t0.date} ({t0} to {t1}) =====")

    for sta in stations:
        try:
            inv = client.get_stations(
                network=network,
                station=sta,
                location="*",
                channel="*",
                starttime=t0,
                endtime=t1,
                level="channel",
            )

            loc, band = pick_band_and_loc(inv)
            if loc is None:
                print(f"{network}.{sta}: no complete ZNE triplet found")
                continue

            # Use '?' wildcard (valid for FDSN), then keep only Z/N/E
            st = client.get_waveforms(
                network=network,
                station=sta,
                location=loc,
                channel=f"{band}?",
                starttime=t0,
                endtime=t1,
                attach_response=False,  # raw counts
            )

            st = st.select(channel=f"{band}[ZNE]")  # ObsPy-side filtering is fine
            st.detrend(type="linear")
            #st.taper(max_percentage=0.05)
            #st.filter("highpass", freq=3)
            channels = {tr.stats.channel for tr in st}
            needed = {f"{band}Z", f"{band}N", f"{band}E"}
            if not needed.issubset(channels):
                print(f"{network}.{sta}: incomplete {band} triplet returned ({sorted(channels)})")
                continue

            st.sort(keys=["channel"])
            print(f"{network}.{sta}  loc='{loc}'  band='{band}'")
            print(st)

            st.plot(
                type="relative",
                equal_scale=False,
                size=(1100, 700),
                method="full",
                suptitle=f"{network}.{sta} raw {band}ZNE ({t0} + 5 min)",
            )

        except Exception as e:
            warnings.warn(f"Failed for {network}.{sta} at {t0}: {type(e).__name__}: {e}")

    t0 += 3600  # next hour
```
