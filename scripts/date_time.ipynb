```python
# Import requirements

import pandas as pd
```

```python
# Load in catalog dataframe
catalog = pd.read_csv('../data/processed_catalog.csv')
```

```python
display(catalog)
```

```python
# Format the date_time column using existing columns
catalog['date_time'] = catalog.apply(
    lambda row: f"{row['year']}-{row['month']:02d}-{row['day']:02d}T{row['hour']:02d}:{row['minute']:02d}:{row['second']:04.1f}Z", 
    axis=1
)

# Display the first few rows to verify the formatting
print(catalog[['year', 'month', 'day', 'hour', 'minute', 'second', 'date_time']].head())
```

```python
# Convert date_time to datetime format
catalog['datetime_obj'] = pd.to_datetime(catalog['date_time'])

# Add 20 seconds to create end_time
catalog['datetime_end'] = catalog['datetime_obj'] + pd.Timedelta(seconds=20)

# Format the end_time in the same format as date_time
catalog['end_time'] = catalog['datetime_end'].dt.strftime('%Y-%m-%dT%H:%M:%S.%fZ').str[:-5] + 'Z'

# Clean up temporary columns
catalog = catalog.drop(['datetime_obj', 'datetime_end'], axis=1)

# Display the first few rows to verify the formatting
print(catalog[['date_time', 'end_time']].head())
```

```python
display(catalog)
```

```python
# date_time and end_time are now in the correct format for ObsPy and get_trace function.

# save the updated catalog to a CSV file
catalog.to_csv('../data/processed_catalog_date_time.csv', index=False)
```
