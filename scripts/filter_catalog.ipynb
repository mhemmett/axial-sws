```python
import obspy
import pickle
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import swspy
```

```python
catalog = pd.read_csv('../data/final_catalog.csv')
```

```python
display(catalog)
```

```python
catalog = catalog.drop(['Unnamed: 0', 'catalog_idx', 'catalog_month', 'catalog_day', 'catalog_hour','catalog_minute', 'catalog_second', 'phase_event_id'], axis=1)
```

```python
display(catalog)
```

```python
catalog = catalog.dropna(ignore_index = True)
```

```python
catalog = catalog.drop('has_data', axis=1)
```

```python
display(catalog)
```

```python
# Check column names
print("Current column names:")
print(catalog.columns.tolist())
```

```python
# Create a new dataframe with renamed columns (remove "catalog_" prefix)
catalog_renamed = catalog.copy()

# Create a mapping dictionary to rename columns
column_mapping = {}
for col in catalog_renamed.columns:
    if col.startswith('catalog_'):
        # Remove "catalog_" prefix
        new_name = col.replace('catalog_', '')
        column_mapping[col] = new_name

print("Column renaming mapping:")
for old_name, new_name in column_mapping.items():
    print(f"'{old_name}' -> '{new_name}'")

# Apply the renaming
catalog_renamed = catalog_renamed.rename(columns=column_mapping)

print(f"\nNew column names:")
print(catalog_renamed.columns.tolist())
```

```python
# Display the new dataframe with renamed columns
print("New dataframe 'catalog_renamed' with cleaned column names:")
display(catalog_renamed)
```

```python
# Filter out earthquakes from year 2014
print(f"Original dataframe shape: {catalog_renamed.shape}")
print(f"Number of earthquakes from 2014: {(catalog_renamed['year'] == 2014).sum()}")

# Create filtered dataframe excluding year 2014
catalog_filtered = catalog_renamed[catalog_renamed['year'] != 2014].copy()

print(f"Filtered dataframe shape: {catalog_filtered.shape}")
print(f"Earthquakes removed: {catalog_renamed.shape[0] - catalog_filtered.shape[0]}")

# Check the year range in the filtered data
print(f"\nYear range in filtered data: {catalog_filtered['year'].min()} to {catalog_filtered['year'].max()}")
print(f"Unique years: {sorted(catalog_filtered['year'].unique())}")
```

```python
display(catalog_filtered)
```

```python
# Filter out rows where p_arrival_time < 0
print(f"Original filtered dataframe shape: {catalog_filtered.shape}")
print(f"Number of rows with p_arrival_time < 0: {(catalog_filtered['p_arrival_time'] < 0).sum()}")

# Create new dataframe excluding rows with p_arrival_time < 0
catalog_p_filtered = catalog_filtered[catalog_filtered['p_arrival_time'] >= 0].copy()

print(f"New dataframe shape after p_arrival_time filter: {catalog_p_filtered.shape}")
print(f"Rows removed: {catalog_filtered.shape[0] - catalog_p_filtered.shape[0]}")

# Check the range of p_arrival_time values
print(f"\nP arrival time range: {catalog_p_filtered['p_arrival_time'].min()} to {catalog_p_filtered['p_arrival_time'].max()}")
print(f"Number of valid P arrival times: {catalog_p_filtered['p_arrival_time'].notna().sum()}")
```

```python
# Save the complete filtered dataframe to CSV
catalog_p_filtered.to_csv('../data/catalog_filtered.csv', index=False)

# Save only the first 100 rows to a separate CSV file
catalog_p_filtered.head(100).to_csv('../data/catalog_filtered_sample.csv', index=False)

print(f"Full dataset saved to '../data/catalog_filtered.csv'")
print(f"Sample of 100 rows saved to '../data/catalog_filtered_sample.csv'")
print(f"Full dataset size: {catalog_p_filtered.shape[0]} rows x {catalog_p_filtered.shape[1]} columns")
print(f"Sample dataset size: 100 rows x {catalog_p_filtered.shape[1]} columns")
```
